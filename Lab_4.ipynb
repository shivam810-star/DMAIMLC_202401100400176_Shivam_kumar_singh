```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, LSTM, GRU, Dense, Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.datasets import imdb

# Load dataset
vocab_size = 10000
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=vocab_size)

# Reduce size (optional for speed)
x_train = x_train[:10000]
y_train = y_train[:10000]
x_test = x_test[:2000]
y_test = y_test[:2000]

# Padding
max_len = 100
x_train = pad_sequences(x_train, maxlen=max_len)
x_test = pad_sequences(x_test, maxlen=max_len)

# Function to build model
def build_model(layer):
    model = Sequential([
        Embedding(vocab_size, 64, input_length=max_len),
        layer(64),
        Dense(1, activation='sigmoid')
    ])
    model.compile(loss='binary_crossentropy',
                  optimizer='adam',
                  metrics=['accuracy'])
    return model

# 🔹 Train RNN
print("\nTraining RNN...")
rnn_model = build_model(SimpleRNN)
rnn_model.fit(x_train, y_train, epochs=2, batch_size=64, verbose=0)
rnn_loss, rnn_acc = rnn_model.evaluate(x_test, y_test, verbose=0)
print("RNN Accuracy:", rnn_acc)

# 🔹 Train LSTM
print("\nTraining LSTM...")
lstm_model = build_model(LSTM)
lstm_model.fit(x_train, y_train, epochs=2, batch_size=64, verbose=0)
lstm_loss, lstm_acc = lstm_model.evaluate(x_test, y_test, verbose=0)
print("LSTM Accuracy:", lstm_acc)

# 🔹 Train GRU
print("\nTraining GRU...")
gru_model = build_model(GRU)
gru_model.fit(x_train, y_train, epochs=2, batch_size=64, verbose=0)
gru_loss, gru_acc = gru_model.evaluate(x_test, y_test, verbose=0)
print("GRU Accuracy:", gru_acc)

# 🔥 Final Comparison
print("\nFinal Comparison:")
print(f"RNN Accuracy  : {rnn_acc:.4f}")
print(f"LSTM Accuracy : {lstm_acc:.4f}")
print(f"GRU Accuracy  : {gru_acc:.4f}")
```
